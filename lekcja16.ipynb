Dataset:

Wymagania:

Narysuj ASCII-histogram funkcją ascii_histogram(x, bins=20, height=10, label="N(50,10)") z tej
lekcji
W podpisie oznacz słownie, gdzie leżą granice ±1σ (np. |---|---[μ-σ]===[μ]===[μ+σ]---|---| )

```python
import math
import numpy as np

rng = np.random.default_rng(seed=42)
x = rng.normal(loc=50, scale=10, size=1000)

# Prawdopodobieństwo -dystrybuanta
def normal_cdf(x, mu=0, sigma=1):
    return 0.5 * (1 + math.erf((x - mu) / (sigma * math.sqrt(2))))

def ascii_histogram(x, bins=20, height=10, label="Histogram"):
    counts, bin_edges = np.histogram(x, bins=bins)
    # print(counts,bin_edges)
    max_count = max(counts)
    
    print(f"\n=== ASCII Histogram: {label} ===")
    for h in range(height, 0, -1):
        row = ""
        for c in counts:
            if (c / max_count) * height >= h:
                row += "█"
            else:
                row += " "
        print(f"{row}")
    print("-" * bins)
    print(f"Min: {bin_edges[0]:.1f} | Max: {bin_edges[-1]:.1f}\n")


empiryczny_procent = ((x >= 40) & (x <= 60)).mean() * 100
teoretyczny_procent = (normal_cdf(60, 50, 10) - normal_cdf(40, 50, 10)) * 100

print(f"Empiryczny % w [40, 60]: {empiryczny_procent:.2f}%")
print(f"Teoretyczny % w [40, 60]: {teoretyczny_procent:.2f}%")

ascii_histogram(x, bins=20, height=10, label="N(50, 10)")
print("Oznaczenie granic: |---|---[μ-σ]===[μ]===[μ+σ]---|---|")
```

Wczytaj syntetyczny dataset kwiatów ( c z Lekcji 15). Sprawdź normalność
rozkładu cechy sepal_length (kolumna 0) testem Jarque-Bera i Q-Q plot.
Dataset:
X, y = make_synthetic_flowers() (150×4, 3 gatunki)


```python
import math
import numpy as np

# Przygotowanie danych syntetycznych kwiatów
def make_synthetic_flowers(seed=42):
    rng = np.random.default_rng(seed)
    
    g1 = rng.normal(loc=[5.0, 3.4, 1.5, 0.2], scale=[0.3, 0.3, 0.2, 0.1], size=(50, 4))
    # print(g1)
    g2 = rng.normal(loc=[6.0, 2.7, 4.2, 1.3], scale=[0.4, 0.3, 0.4, 0.2], size=(50, 4))
    g3 = rng.normal(loc=[6.5, 3.0, 5.5, 2.0], scale=[0.5, 0.3, 0.5, 0.2], size=(50, 4))
    X = np.vstack([g1, g2, g3])
    y = np.array([0]*50 + [1]*50 + [2]*50)
    # print(y)
    return X, y

#cCumulative Distribution Function -Mając konkretną wartość x, 
# jakie jest prawdopodobieństwo (p), że wylosowana zmienna będzie od niej mniejsza lub równa?
def normal_cdf(x, mu=0, sigma=1):
    return 0.5 * (1 + math.erf((x - mu) / (sigma * math.sqrt(2))))
#Percent Point Function -f.kwantyli -Jaka wartość x odcina w rozkładzie wskazany odsetek/prawdopodobieństwo (p)?
def normal_ppf(p, mu=0, sigma=1, tol=1e-6):
    
    low, high = -10.0, 10.0
    while high - low > tol:
        mid = (low + high) / 2.0
        if normal_cdf(mid) < p:
            low = mid
        else:
            high = mid
    return mu + sigma * ((low + high) / 2.0)

def jarque_bera_manual(x):
    n = len(x)
    mean = np.mean(x)
    std = np.std(x, ddof=0)
    s = np.mean(((x - mean) / std) ** 3)
    k = np.mean(((x - mean) / std) ** 4) - 3
    jb = (n / 6) * (s**2 + (k**2) / 4)
    p_val = math.exp(-jb / 2)
    return jb, p_val

def ascii_qq_plot(x, rows=10, cols=30):
    sorted_x = np.sort(x)
    n = len(x)
    p = (np.arange(1, n + 1) - 0.5) / n
    theoretical_q = np.array([normal_ppf(pi) for pi in p])
    
    # Standaryzacja
    sample_q = (sorted_x - np.mean(sorted_x)) / np.std(sorted_x)
    
    grid = [[" " for _ in range(cols)] for _ in range(rows)]
    for t, s in zip(theoretical_q, sample_q):
        r = int(np.clip((1 - (s + 3) / 6) * (rows - 1), 0, rows - 1))
        c = int(np.clip(((t + 3) / 6) * (cols - 1), 0, cols - 1))
        grid[r][c] = "*"
        
    print("\n--- ASCII Q-Q Plot ---")
    for row in grid:
        print("".join(row))

X, y = make_synthetic_flowers()
sepal_length = X[:, 0]


jb_all, p_all = jarque_bera_manual(sepal_length)
print(f"Cała populacja sepal_length: JB = {jb_all:.4f}, p-value = {p_all:.4f}")
ascii_qq_plot(sepal_length)


for species_id in range(3):
    sub_x = sepal_length[y == species_id]
    jb, p = jarque_bera_manual(sub_x)
    print(f"Gatunek {species_id}: JB = {jb:.4f}, p-value = {p:.4f}")
```

Rozkład cechy sepal_length nie jest normalny dla całej populacji, co widać również po schodkowym układzie punktów na wykresie Q-Q Plot.   Natomiast wewnątrz każdego z pojedynczych gatunków rozkład jest normalny, ponieważ dla każdego z nich p{-value} > 0.05$ 

Dla datasetu kwiatów ( make_synthetic_flowers ) oblicz korelację Pearsona między kolumnami 2
( petal_length ) a 3 ( petal_width ). Narysuj ASCII scatter plot z linią regresji.
Dataset:
make_synthetic_flowers()
Wymagania:
Współczynnik: pearson_corr(x, y) z Przykładu 2 Sekcji 2
p-value z aproksymacji normalnej: corr_pvalue_approx(r, n) z tej samej sekcji
ASCII scatter plot: ascii_scatter(x, y, title="petal_length vs petal_width", rows=20, cols=60)
(funkcja sama dorysuje linię regresji przez np.polyfit )
Jeżeli chcesz obliczyć współczynniki regresji ręcznie: a = r * std_y/std_x , b = mean_y - a*mean_x
Interpretacja siły i kierunku korelacji (|r| > 0.7 = silna)


```python
import math
import numpy as np

def make_synthetic_flowers(seed=42):
    rng = np.random.default_rng(seed)
    g1 = rng.normal(loc=[5.0, 3.4, 1.5, 0.2], scale=[0.3, 0.3, 0.2, 0.1], size=(50, 4))
    g2 = rng.normal(loc=[6.0, 2.7, 4.2, 1.3], scale=[0.4, 0.3, 0.4, 0.2], size=(50, 4))
    g3 = rng.normal(loc=[6.5, 3.0, 5.5, 2.0], scale=[0.5, 0.3, 0.5, 0.2], size=(50, 4))
    return np.vstack([g1, g2, g3]), np.array([0]*50 + [1]*50 + [2]*50)

def pearson_corr(x, y):
    mx, my = np.mean(x), np.mean(y)
    num = np.sum((x - mx) * (y - my))
    den = np.sqrt(np.sum((x - mx)**2) * np.sum((y - my)**2))
    return num / den

def normal_cdf(x, mu=0, sigma=1):
    return 0.5 * (1 + math.erf((x - mu) / (sigma * math.sqrt(2))))

def corr_pvalue_approx(r, n):
    t = r * math.sqrt(n - 2) / math.sqrt(1 - r**2)
    return 2 * (1 - normal_cdf(abs(t)))

def ascii_scatter(x, y, title="", rows=15, cols=40):
    grid = [[" " for _ in range(cols)] for _ in range(rows)]
    x_min, x_max = np.min(x), np.max(x)
    y_min, y_max = np.min(y), np.max(y)
    

    a, b = np.polyfit(x, y, 1)
    

    for xi, yi in zip(x, y):
        c = int(np.clip(((xi - x_min) / (x_max - x_min)) * (cols - 1), 0, cols - 1))
        r = int(np.clip((1 - (yi - y_min) / (y_max - y_min)) * (rows - 1), 0, rows - 1))
        grid[r][c] = "*"
        
    print(f"\n--- Scatter Plot: {title} ---")
    for row in grid:
        print("".join(row))

X, _ = make_synthetic_flowers()
petal_length = X[:, 2]
petal_width = X[:, 3]

r = pearson_corr(petal_length, petal_width)
p_val = corr_pvalue_approx(r, len(petal_length))

print(f"Współczynnik korelacji Pearsona (r): {r:.4f}")
print(f"Aproksymowane p-value: {p_val:.4e}")

ascii_scatter(petal_length, petal_width, title="petal_length vs petal_width")
```

Dla datasetu California Housing (generator make_california_housing() z Lekcji 12) narysuj ASCIIheatmap macierzy korelacji. Zidentyfikuj 3 pary cech o najwyższej korelacji (poza targetem MedHouseVal ).

```python
import numpy as np
import pandas as pd

def make_california_housing(n_samples=2000, seed=42):
    rng = np.random.default_rng(seed)
    cols = ['MedInc', 'HouseAge', 'AveRooms', 'AveBedrms', 'Population', 'AveOccup', 'Latitude', 'Longitude', 'MedHouseVal']
    data = rng.normal(size=(n_samples, len(cols)))
    # print(data)
    data[:, 3] = data[:, 2] * 0.8 + rng.normal(scale=0.2, size=n_samples) # AveBedrms vs AveRooms
    data[:, 7] = -data[:, 6] * 0.9 + rng.normal(scale=0.1, size=n_samples) # Longitude vs Latitude
    data[:, 8] = data[:, 0] * 0.7 + rng.normal(scale=0.3, size=n_samples)  # MedHouseVal vs MedInc
    return pd.DataFrame(data, columns=cols)


def ascii_heatmap_signed(matrix, labels):
    print("\n=== ASCII Heatmap (Znak i Siła Korelacji) ===")
    print("Legenda: '-' (ujemna) | '.' (brak) | '+' (dodatnia) | '@' (silna dodatnia)")
    
    for i, row in enumerate(matrix):
        row_str = ""
        for val in row:
            if val > 0.7:
                row_str += "@"   # Bardzo silna dodatnia
            elif val > 0.3:
                row_str += "+"   # Umiarkowana dodatnia
            elif val < -0.7:
                row_str += "#"   # Bardzo silna ujemna
            elif val < -0.3:
                row_str += "-"   # Umiarkowana ujemna
            else:
                row_str += "."   # Brak korelacji / blisko zera
                
        print(f"{labels[i][:8]:>8} | {row_str}")

df = make_california_housing(n_samples=2000)
C = np.corrcoef(df.values.T)

ascii_heatmap_signed(C, df.columns)

#  top-3 par cech 
feature_cols = [c for c in df.columns if c != 'MedHouseVal']
pairs = []

for i in range(len(feature_cols)):
    for j in range(i + 1, len(feature_cols)):
        idx_i = df.columns.get_loc(feature_cols[i])
        idx_j = df.columns.get_loc(feature_cols[j])
        corr_val = C[idx_i, idx_j]
        pairs.append((feature_cols[i], feature_cols[j], corr_val, abs(corr_val)))
        # print(pairs)

pairs.sort(key=lambda x: x[3], reverse=True)


for p in pairs[:3]:
    print(f"- {p[0]} vs {p[1]}: r = {p[2]:.4f}")
```

Dla datasetu Titanic przetestuj hipotezę, że średni wiek pasażerów wynosi 30 lat (H₀: μ = 30).

Titanic: df =
pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
(legalne od L11), ages = df['Age'].dropna().values
Wymagania:
Statystyka t: t = (x.mean() - 30) / (x.std(ddof=1) / np.sqrt(n))
p-value: dla dużego n (>30) aproksymacja normalna — p = 2 * (1 - normal_cdf(abs(t), 0, 1))
Wniosek: jeśli p < 0.05 → odrzucamy H₀
ASCII-oznaczenie pozycji x̄ względem μ₀ na prostej ( ├───x̄───μ₀───┤ )

```python
import math
import numpy as np
import pandas as pd

def normal_cdf(x, mu=0, sigma=1):
    return 0.5 * (1 + math.erf((x - mu) / (sigma * math.sqrt(2))))

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)
ages = df['Age'].dropna().values

n = len(ages)
mu_0 = 30
x_bar = ages.mean()
s = ages.std(ddof=1)

t_stat = (x_bar - mu_0) / (s / np.sqrt(n))
p_val = 2 * (1 - normal_cdf(abs(t_stat), 0, 1))

print(f"Średnia wieku x̄: {x_bar:.2f}")
print(f"Statystyka t: {t_stat:.4f}")
print(f"p-value: {p_val:.4f}")

if p_val < 0.05:
    print("Wniosek: Odrzucamy H0 na poziomie istotności alpha=0.05.")
else:
    print("Wniosek: Brak podstaw do odrzucenia H0.")


print("\nPozycja x̄ względem μ₀ na prostej:")
if x_bar < mu_0:
    print("├───x̄───μ₀───┤")
else:
    print("├───μ₀───x̄───┤")
```

7. Porównaj średni wiek pasażerów Titanica, którzy przeżyli vs nie przeżyli. Czy różnica jest istotna
statystycznie?
Dataset:
Titanic (ta sama metoda wczytania co w Zad. 6)
Wymagania:
Welch's t-test ręcznie (wzory z Przykładu 1 Sekcji 3)
p-value przez permutation_pvalue(a, b, n_perm=2000) z Przykładu 1 Sekcji 3
ASCII-boxplot porównujący wiek: ascii_boxplot_two(age_survived, age_died)
Wniosek z interpretacją p-value i Cohena's d


```python
import math
import numpy as np
import pandas as pd

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

age_survived = df[df['Survived'] == 1]['Age'].dropna().values
age_died = df[df['Survived'] == 0]['Age'].dropna().values



def permutation_pvalue(a, b, n_perm=2000, seed=42):
    rng = np.random.default_rng(seed)
    diff_obs = abs(np.mean(a) - np.mean(b))
    combined = np.concatenate([a, b])
    n_a = len(a)
    
    count = 0
    for _ in range(n_perm):
        rng.shuffle(combined)
        diff_perm = abs(np.mean(combined[:n_a]) - np.mean(combined[n_a:]))
        if diff_perm >= diff_obs:
            count += 1
    return count / n_perm

def ascii_boxplot_two(a, b, label1="Survived", label2="Died"):
    print(f"\n--- Porównanie średnich ---")
    print(f"{label1:>10}: Średnia = {np.mean(a):.2f}, Odchylenie = {np.std(a):.2f}")
    print(f"{label2:>10}: Średnia = {np.mean(b):.2f}, Odchylenie = {np.std(b):.2f}")



# Welch's t-test -test Welcha stosuje się wtedy, gdy porównujemy dwie grupy, które mogą mieć różne liczności oraz różne wariancje 
n1, n2 = len(age_survived), len(age_died)
m1, m2 = np.mean(age_survived), np.mean(age_died)
v1, v2 = np.var(age_survived, ddof=1), np.var(age_died, ddof=1)

t_welch = (m1 - m2) / np.sqrt(v1/n1 + v2/n2)
p_perm = permutation_pvalue(age_survived, age_died, n_perm=2000)

# Cohen's d
s_pooled = np.sqrt(((n1 - 1)*v1 + (n2 - 1)*v2) / (n1 + n2 - 2))
cohens_d = (m1 - m2) / s_pooled

print(f"Welch t-stat: {t_welch:.4f}")
print(f"Permutation p-value: {p_perm:.4f}")
print(f"Cohen's d: {cohens_d:.4f}")

ascii_boxplot_two(age_survived, age_died)
```

Różnica średniego wieku między pasażerami, którzy przeżyli (ok. 28.3 lat), a tymi, którzy zginęli (ok. 30.6 lat), jest statystycznie istotna ($p\text{-value} = 0.0405 < 0.05$), co oznacza, że osoby, które przeżyły, były średnio nieco młodsze. Jednakże miara wielkości efektu Cohena ($d = -0.16$) wskazuje, że efekt ten jest bardzo słaby w ujęciu praktycznym.

8. Przetestuj, czy przeżycie na Titanicu zależy od klasy podróży (Pclass). Narysuj stacked bar chart z procentami.
Dataset:
Titanic
Wymagania:
Tabela kontyngencji przez pd.crosstab(df['Pclass'], df['Survived'])
Statystyka χ²: ręcznie sum((O - E)**2 / E) (patrz Przykład 2 Sekcji 3)
p-value przez chi2_monte_carlo_pvalue(observed, n_sim=3000) z Przykładu 2 Sekcji 3
Stacked bar chart funkcją ascii_stacked_bars(table) z Przykładu 2 Sekcji 3
Cramer's V: V = sqrt(chi2 / (n * min(r-1, c-1)))


```python
import math
import numpy as np
import pandas as pd


url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)


def chi2_monte_carlo_pvalue(observed, n_sim=3000, seed=42):
    rng = np.random.default_rng(seed)
    row_sums = observed.sum(axis=1)
    col_sums = observed.sum(axis=0)
    total = observed.sum()
    
    expected = np.outer(row_sums, col_sums) / total
    chi2_obs = np.sum((observed - expected)**2 / expected)
    
    probs = col_sums / total
    count = 0
    for _ in range(n_sim):
        sim_table = np.zeros_like(observed)
        for i, r_sum in enumerate(row_sums):
            counts = rng.multinomial(r_sum, probs)
            sim_table[i] = counts
        chi2_sim = np.sum((sim_table - expected)**2 / expected)
        if chi2_sim >= chi2_obs:
            count += 1
    return chi2_obs, count / n_sim


def ascii_stacked_bars(table):
    print("\n--- ASCII Stacked Bar Chart (Procent przeżycia w klasach) ---")
    props = table.div(table.sum(axis=1), axis=0) * 100
    for idx, row in props.iterrows():
        surv_len = int(row[1] // 5)
        died_len = int(row[0] // 5)
        bar = "█" * surv_len + "░" * died_len
        print(f"Klasa {idx}: [{bar}] (Przeżyło: {row[1]:.1f}%)")


ct = pd.crosstab(df['Pclass'], df['Survived']).values
chi2_stat, p_val = chi2_monte_carlo_pvalue(ct, n_sim=3000)

n = ct.sum()
r, c = ct.shape
cramers_v = math.sqrt(chi2_stat / (n * min(r - 1, c - 1)))

print(f"Chi2 Statystyka: {chi2_stat:.4f}")
print(f"p-value (Monte Carlo): {p_val:.4f}")
print(f"Cramer's V: {cramers_v:.4f}")

ascii_stacked_bars(pd.crosstab(df['Pclass'], df['Survived']))
```

Chi2 Statystyka (102.89): Bardzo wysoka wartość statystyki testowej wskazuje na dużą rozbieżność między stanem faktycznym a sytuacją, w której klasa nie miałaby żadnego wpływu na przeżycie.p-value (0.0000): Ponieważ wartość p jest mniejsza niż jakikolwiek standardowy poziom istotności (np. alpha = 0.05), kategorycznie odrzucamy hipotezę zerową H0).Wniosek: Przeżycie na Titanicu nie było dziełem przypadku – istniała statystycznie istotna zależność między klasą biletu a szansą na przeżycie.


Cramer's V (0.3398): Miara ta określa siłę powiązania między zmiennymi w skali od 0 do 1. Wynik w okolicach 0.34 wskazuje na umiarkowany do silnego związek pomiędzy klasą podróży a przeżyciem.

Wyniki potwierdzają historyczną regułę katastrofy Titanica: im wyższa klasa podróży, tym większe były szanse na przeżycie (co wynikało m.in. z lokalizacji kabin bliżej pokładu łodzi ratunkowych oraz pierwszeństwa w ewakuacji).

9. Symuluj 1000 eksperymentów rzutu 20 monetami (p=0.5). Porównaj histogram wyników z teoretycznym
PMF B(20, 0.5).
Dataset:
rng = np.random.default_rng(seed=42) , sim = rng.binomial(n=20, p=0.5, size=1000)
Wymagania:
Histogram symulacji przez np.bincount(sim, minlength=21)
Teoretyczny PMF przez funkcję binom_pmf(k, n, p) z Przykładu 3 Sekcji 1 (lub [math.comb(20,k) *
0.5**20 for k in range(21)] )
ASCII bar chart dwuprzebiegowy: linia symulacji █ i linia teorii · (użyj ascii_barplot_pmf z
Przykładu 3 — dwa razy, z różnymi znacznikami)
Porównanie średniej empirycznej ( sim.mean() ) z teoretyczną ( n*p = 10 )

```python

import math
import numpy as np

def binom_pmf(k, n, p):
    return math.comb(n, k) * (p**k) * ((1 - p)**(n - k))

rng = np.random.default_rng(seed=42)
sim = rng.binomial(n=20, p=0.5, size=1000)

# zlicza, ile razy wystąpił każdy możliwy wynik
emp_counts = np.bincount(sim, minlength=21) 
emp_pmf = emp_counts / len(sim)
# Dla każdego możliwego wyniku k (od 0 do 20) obliczamy idealne, teoretyczne prawdopodobieństwo wystąpienia takiego wyniku matematycznego.
theo_pmf = [binom_pmf(k, 20, 0.5) for k in range(21)]

print("=== ASCII Bar Chart: Symulacja (█) vs Teoria (·) ===")
max_val = max(max(emp_pmf), max(theo_pmf))
height = 10

for h in range(height, 0, -1):
    row = ""
    for k in range(21):
        emp_h = (emp_pmf[k] / max_val) * height
        theo_h = (theo_pmf[k] / max_val) * height
        
        if abs(emp_h - h) < 0.5 and abs(theo_h - h) < 0.5:
            row += "░"
        elif emp_h >= h:
            row += "█"
        elif theo_h >= h:
            row += "·"
        else:
            row += " "
    print(f"{row}")
print("-" * 21)
print("k: " + "".join([str(k % 10) for k in range(21)]))

print(f"\nŚrednia empiryczna: {sim.mean():.4f}")
print(f"Średnia teoretyczna (n*p): {20 * 0.5:.4f}")
```

16. Dla datasetu make_synthetic_wine() zidentyfikuj silnie skorelowane pary cech (|r| > 0.7) i usuń je
iteracyjnie metodą "greedy correlation pruning". Porównaj macierz korelacji przed i po.
Dataset:
df = make_synthetic_wine(n_samples=1600, seed=42) , cols = [c for c in df.columns if c !=
"quality"] , X = df[cols].values
> [!note] > Klasyczne VIF (Variance Inflation Factor) wymaga regresji liniowej, którą poznasz formalnie w
Lekcji 20. Do tego czasu używamy prostszej, ale skutecznej heurystyki opartej na macierzy korelacji — w
praktyce wyniki pokrywają się z VIF dla większości datasetów.
Wymagania:
Oblicz C = np.corrcoef(X.T) i ASCII-heatmap funkcją ascii_heatmap(C, labels=cols)
Algorytm iteracyjny:
1. Znajdź parę (i, j) z największym |C[i,j]| (poza diagonalą)
2. Jeśli |C[i,j]| > 0.7 : usuń jedną z cech (tę o większej średniej korelacji z pozostałymi)
3. Przelicz macierz, wróć do 1
4. Zatrzymaj, gdy wszystkie |C| <= 0.7
Lista usuniętych cech
ASCII-heatmap po usunięciu
Oczekiwany rezultat:
2-4 cechy usunięte (typowo pH/acidity pair)
Finalna macierz ma wszystkie |r| ≤ 0.7

```python
import numpy as np
import pandas as pd

def make_synthetic_wine(n_samples=1600, seed=42):
    rng = np.random.default_rng(seed)
    cols = [
        'fixed_acidity', 'volatile_acidity', 'citric_acid', 
        'residual_sugar', 'chlorides', 'free_sulfur_dioxide', 
        'total_sulfur_dioxide', 'density', 'pH', 'sulphates', 
        'alcohol', 'quality'
    ]
    data = rng.normal(size=(n_samples, len(cols)))
    
   
    data[:, 8] = -data[:, 0] * 0.85 + rng.normal(scale=0.1, size=n_samples) # pH vs fixed_acidity
    data[:, 6] = data[:, 5] * 0.78 + rng.normal(scale=0.2, size=n_samples) # total_sulfur vs free_sulfur
    
    return pd.DataFrame(data, columns=cols)


def ascii_heatmap(matrix, labels):
    print("\n=== ASCII Heatmap (Znak i Siła Korelacji) ===")
    print("Legenda: '-' (ujemna) | '.' (brak) | '+' (dodatnia) | '@' (silna dodatnia)")
    
    for i, row in enumerate(matrix):
        row_str = ""
        for val in row:
            if val > 0.7:
                row_str += "@"   # Bardzo silna dodatnia
            elif val > 0.3:
                row_str += "+"   # Umiarkowana dodatnia
            elif val < -0.7:
                row_str += "#"   # Bardzo silna ujemna
            elif val < -0.3:
                row_str += "-"   # Umiarkowana ujemna
            else:
                row_str += "."   # Brak korelacji / blisko zera
                
        print(f"{labels[i][:15]:>15} | {row_str}")

# Wczytanie danych i wykluczenie zmiennej celu ('quality')[cite: 3]
df = make_synthetic_wine(n_samples=1600, seed=42)
cols = [c for c in df.columns if c != "quality"]
X = df[cols].values

# Obliczenie początkowej macierzy korelacji i wyświetlenie mapy ASCII[cite: 3]
C = np.corrcoef(X.T)
print("=== Macierz korelacji PRZED redukcją ===")
ascii_heatmap(C, labels=cols)

#Algorytm zachłannego usuwania korelacji (Greedy Correlation Pruning)
current_cols = list(cols)
removed_features = []

while True:
    X_curr = df[current_cols].values
    C_curr = np.corrcoef(X_curr.T)
    
    # Wyzerowanie przekątnej (korelacja cechy z samą sobą to 1.0)[cite: 3]
    np.fill_diagonal(C_curr, 0)
    
    # Znajdź największą wartość bezwzględną korelacji[cite: 3]
    max_corr = np.max(np.abs(C_curr))
    
    # Warunek stopu: jeśli największa korelacja <= 0.7, kończymy[cite: 3]
    if max_corr <= 0.7:
        break
        
    # Znajdź indeksy (i, j) tej maksymalnej korelacji[cite: 3]
    i, j = np.unravel_index(np.argmax(np.abs(C_curr)), C_curr.shape)
    
    col_i = current_cols[i]
    col_j = current_cols[j]
    
    # Kryterium wyboru: usuwamy tę cechę, która ma wyższą średnią korelację z resztą[cite: 3]
    mean_corr_i = np.mean(np.abs(C_curr[i, :]))
    mean_corr_j = np.mean(np.abs(C_curr[j, :]))
    
    to_remove = col_i if mean_corr_i > mean_corr_j else col_j
    removed_features.append(to_remove)
    current_cols.remove(to_remove)
    print(f"\nUsuwamy cechę: '{to_remove}' (powiązaną z '{col_i if to_remove != col_i else col_j}', max |r| = {max_corr:.4f})")

#  Obliczenie macierzy po usunięciu cech
X_final = df[current_cols].values
C_final = np.corrcoef(X_final.T)

print(f"\n\nLista usuniętych cech: {removed_features}")
print(f"Liczba usuniętych cech: {len(removed_features)} (oczekiwane: 2-4)")
print("\n=== Macierz korelacji PO REDUKCJI (wszystkie |r| <= 0.7) ===")
ascii_heatmap(C_final, labels=current_cols)
```
