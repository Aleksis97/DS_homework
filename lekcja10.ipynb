```python
# Utwórz tablicę NumPy zawierającą liczby od 1 do 100. Oblicz i wypisz:
# Sumę wszystkich elementów
# Średnią arytmetyczną
# Medianę
# Wartość maksymalną i minimalną
# Wymagania:
# Użyj np.arange() do stworzenia tablicy
# Użyj funkcji agregujących NumPy ( np.sum , np.mean , ...

import numpy as np

array = np.arange(1, 101)


total_sum = np.sum(array)
mean_val = np.mean(array)
median_val = np.median(array)
min_val = np.min(array)
max_val = np.max(array)

print(f"Tablica: {array}")
print(f"Suma: {total_sum}")
print(f"Średnia arytmetyczna: {mean_val}")
print(f"Mediana: {median_val}")
print(f"Wartość minimalna: {min_val}")
print(f"Wartość maksymalna: {max_val}")
```

```python
before = np.array([100, 250, 75, 320, 180, 450, 90])
after = np.array([ 80, 200, 60, 280, 150, 400, 75])
# Oblicz i wypisz:
roznice=before-after
print(roznice)
# Różnicę między cenami (oszczędności)
# Procent rabatu dla każdego produktu
# Średnią oszczędność
# Wymagania:
# Użyj operacji wektorowych (bez pętli)
# Wynik zaokrąglij do 2 miejsc po przecinku


procent_rabatu = np.round((roznice / before) * 100, 2)

srednia_oszczednost = round(np.mean(roznice), 2)

print(f"Różnice w cenach (oszczędności): {roznice}")
print(f"Procent rabatu (%): {procent_rabatu}")
print(f"Średnia oszczędność: {srednia_oszczednost}")
```

```python
# Dana tablica reprezentująca sprzedaż 4 produktów w 7 dni:
np.random.seed(1)
sales = np.random.randint(10, 100, size=(4, 7))
# Oblicz i wypisz:
# Średnią sprzedaż per produkt (axis=1)
# Średnią sprzedaż per dzień (axis=0)
# Dzień z najwyższą łączną sprzedażą ( np.argmax )
# Produkt z najniższą średnią sprzedażą
avg_produkt=np.mean(sales, axis=1)
avg_dzien=np.mean(sales, axis=0)
dzien_najwieksza_Sprzedaz=np.argmax(np.sum(sales, axis=0))
produkt_najwieksza_sprzedaz=np.argmin(avg_produkt)
print(sales)
print(avg_produkt)
print(avg_dzien)
print(dzien_najwieksza_Sprzedaz)
print(produkt_najwieksza_sprzedaz)





# Wypisanie wyników
print(f"Tablica sprzedaży:\n{sales}\n")
print(f"Średnia sprzedaż per produkt: {avg_produkt}")
print(f"Średnia sprzedaż per dzień: {avg_dzien}")
print(f"Dzień z najwyższą łączną sprzedażą (indeks 0-6): {dzien_najwieksza_Sprzedaz}")
print(f"Produkt z najniższą średnią sprzedażą (indeks 0-3): {produkt_najwieksza_sprzedaz}")

```

```python

matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])

print("Oryginalna macierz:\n", matrix)

# 1. Test flatten() - tworzy KOPIĘ
flat = matrix.flatten()
flat[0] = 999  # Modyfikujemy kopię

# 2. Test ravel() - tworzy WIDOK (view)
rav = matrix.ravel()
rav[1] = 888  # Modyfikujemy widok

print("\n--- Po modyfikacjach ---")
print("Zmieniony flat (flatten tworzy kopię, oryginał bez zmian):\n", flat)
print("Zmieniony rav (ravel tworzy widok, wpływa na oryginał):\n", rav)
print("Oryginalna macierz po modyfikacjach (element [0,1] zmienił się na 888):\n", matrix)
```

```python
import numpy as np


np.random.seed(42)
data = np.concatenate([
    np.random.randn(100) * 10 + 50,
    np.array([150, -50, 200, -100, 180])
])
np.random.shuffle(data)


mean_before = np.mean(data)
std_before = np.std(data)
standardized_before = (data - mean_before) / std_before


mask = np.abs(standardized_before) > 3.0
clean_data = data[~mask]


mean_after = np.mean(clean_data)
std_after = np.std(clean_data)
standardized_after = (clean_data - mean_after) / std_after


def print_text_histogram(arr, bins=8):
    counts, edges = np.histogram(arr, bins=bins)
    max_count = counts.max() if counts.max() > 0 else 1
    for i in range(len(counts)):
        bar = "#" * int(counts[i] * 30 / max_count)
        print(f"{edges[i]:6.1f} to {edges[i+1]:6.1f} | {bar} {counts[i]} )")

# Statystyki PRZED
print("--- PRZED USUNIĘCIEM OUTLIERÓW ---")
print(f"Liczba elementów: {len(data)}")
print(f"Średnia (mean): {mean_before:.2f}")
print(f"Odchylenie standardowe (std): {std_before:.2f}")
print(f"Wartość minimalna (min): {np.min(data):.2f}")
print(f"Wartość maksymalna (max): {np.max(data):.2f}")
print("\nHistogram tekstowy (przed):")
print_text_histogram(data)

print("\n" + "=" * 40 + "\n")

# Statystyki PO
print("--- PO USUNIĘCIEM OUTLIERÓW ---")
print(f"Liczba elementów: {len(clean_data)}")
print(f"Średnia (mean): {mean_after:.2f}")
print(f"Odchylenie standardowe (std): {std_after:.2f}")
print(f"Wartość minimalna (min): {np.min(clean_data):.2f}")
print(f"Wartość maksymalna (max): {np.max(clean_data):.2f}")
print("\nHistogram tekstowy (po):")
print_text_histogram(clean_data)
```
