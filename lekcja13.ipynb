```python
# Utwórz dwie tabele: customers (customer_id, name, city) i orders (order_id, customer_id, amount).
# Wymagania:
# Wykonaj INNER JOIN łączący klientów z zamówieniami
# Wyświetl nazwę klienta, miasto i kwotę zamówienia
# Posortuj wyniki według kwoty malejąco


import sqlite3
import pandas as pd

conn = sqlite3.connect('product.db')
cursor = conn.cursor()


cursor.execute('''
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    city TEXT NOT NULL
)
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_id INTEGER NOT NULL,
    amount REAL NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES customers (customer_id)
)
''')

sample_customers = [
    ('Anna Nowak', 'Warszawa'),
    ('Jan Kowalski', 'Kraków'),
    ('Katarzyna Wiśniewska', 'Wrocław'),
    ('Piotr Zieliński', 'Poznań')
]
cursor.executemany('INSERT INTO customers (name, city) VALUES (?, ?)', sample_customers)

sample_orders = [
    (1, 1500.50),
    (2, 350.00),
    (1, 2200.00),
    (3, 499.99),
    (4, 89.00),
    (2, 1250.75)
]
cursor.executemany('INSERT INTO orders (customer_id, amount) VALUES (?, ?)', sample_orders)

conn.commit()

query = '''
SELECT 
    c.name, 
    c.city, 
    o.amount
FROM customers c
INNER JOIN orders o ON c.customer_id = o.customer_id
ORDER BY o.amount DESC
'''

df = pd.read_sql_query(query, conn)
print(df)

conn.close()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('product.db')

query_group_by = '''
SELECT 
    c.name AS customer_name,
    COUNT(o.order_id) AS liczba_zamowien,
    SUM(o.amount) AS lączna_kwota,
    AVG(o.amount) AS srednia_kwota
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.name
'''

df_group = pd.read_sql_query(query_group_by, conn)
print(df_group)

conn.close()
```

```python
import sqlite3
import time
import numpy as np
import pandas as pd


conn = sqlite3.connect('product.db')
cursor = conn.cursor()


cursor.execute('''
CREATE TABLE IF NOT EXISTS products (
    product_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    category TEXT NOT NULL,
    price REAL NOT NULL,
    stock INTEGER NOT NULL
)
''')

cursor.execute('DELETE FROM products')
conn.commit()

num_records = 10000
categories = np.array(['Elektronika', 'Odzież', 'Dom', 'Sport', 'Książki'])

np.random.seed(42)
gen_categories = np.random.choice(categories, size=num_records)
# print(gen_categories)
gen_prices = np.round(np.random.uniform(10.0, 5000.0, size=num_records), 2)
gen_stock = np.random.randint(1, 100, size=num_records)
gen_names = [f"Produkt_{i}" for i in range(num_records)]

data_to_insert = list(zip(gen_names, gen_categories, gen_prices, gen_stock))
cursor.executemany('INSERT INTO products (name, category, price, stock) VALUES (?, ?, ?, ?)', data_to_insert)
conn.commit()

search_category = 'Elektronika'
query = "SELECT * FROM products WHERE category = ?"


times_no_index = []
for _ in range(5):
    start = time.perf_counter()
    cursor.execute(query, (search_category,))
    cursor.fetchall()
    end = time.perf_counter()
    times_no_index.append(end - start)

avg_time_no_index = np.mean(times_no_index)
print(f"Średni czas BEZ indeksu: {avg_time_no_index:.6f} sekund")

explain_no_idx = pd.read_sql_query(f"EXPLAIN QUERY PLAN {query.replace('?', repr(search_category))}", conn)
print(explain_no_idx)

cursor.execute('CREATE INDEX IF NOT EXISTS idx_products_category ON products (category)')
conn.commit()

times_with_index = []
for _ in range(5):
    start = time.perf_counter()
    cursor.execute(query, (search_category,))
    cursor.fetchall()
    end = time.perf_counter()
    times_with_index.append(end - start)

avg_time_with_index = np.mean(times_with_index)
print(f"Średni czas Z INDEKSEM: {avg_time_with_index:.6f} sekund")

explain_with_idx = pd.read_sql_query(f"EXPLAIN QUERY PLAN {query.replace('?', repr(search_category))}", conn)
print("Plan zapytania (Z indeksem):")
print(explain_with_idx)

print("=== 3. PORÓWNANIE ===")
speedup = avg_time_no_index / avg_time_with_index if avg_time_with_index > 0 else 0
print(f"Czas BEZ indeksu: {avg_time_no_index:.6f} s")
print(f"Czas Z indeksem:  {avg_time_with_index:.6f} s")
print(f"Przyspieszenie:   ~{speedup:.2f}x szybsze")

conn.close()
```

```python
import sqlite3
import pandas as pd


conn = sqlite3.connect('product.db')
cursor = conn.cursor()


cursor.execute('''
CREATE TABLE IF NOT EXISTS products (
    product_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    category TEXT NOT NULL,
    price REAL NOT NULL,
    stock INTEGER NOT NULL
)
''')


sample_products = [
    ('Myszka bezprzewodowa', 'Elektronika', 120.00, 15),
    ('Klawiatura mechaniczna', 'Elektronika', 450.00, 8),
    ('Monitor 27 cali', 'Elektronika', 1200.00, 4),
    ('Biurko regulowane', 'Meble', 1800.00, 2),
    ('Fotel biurowy', 'Meble', 950.00, 5),
    ('Książka o Pythonie', 'Książki', 90.00, 30),
    ('Lampa biurkowa', 'Dom', 150.00, 12)
]


cursor.executemany('INSERT INTO products (name, category, price, stock) VALUES (?, ?, ?, ?)', sample_products)
conn.commit()


query = '''
SELECT 
    name, 
    price,
    (price - (SELECT AVG(price) FROM products)) AS roznica_od_sredniej,
    ((price - (SELECT AVG(price) FROM products)) / (SELECT AVG(price) FROM products)) * 100 AS procent_powyzej_sredniej
FROM products
WHERE price > (SELECT AVG(price) FROM products)
ORDER BY price DESC
'''

df_subquery = pd.read_sql_query(query, conn)

print(df_subquery)

conn.close()
```

```python
import sqlite3
import pandas as pd
import numpy as np

conn = sqlite3.connect('product.db')
cursor = conn.cursor()

cursor.execute('DROP TABLE IF EXISTS cohort_orders')
cursor.execute('DROP TABLE IF EXISTS cohort_customers')

cursor.execute('''
CREATE TABLE cohort_customers (
    customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    registration_date TEXT NOT NULL
)
''')

cursor.execute('''
CREATE TABLE cohort_orders (
    order_id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_id INTEGER NOT NULL,
    order_date TEXT NOT NULL,
    amount REAL NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES cohort_customers (customer_id)
)
''')

np.random.seed(42)
n_customers = 50

reg_dates = pd.date_range(start='2025-01-01', end='2026-03-01', periods=n_customers)
customers_data = [(f"Klient_{i}", d.strftime('%Y-%m-%d')) for i, d in enumerate(reg_dates)]
cursor.executemany('INSERT INTO cohort_customers (name, registration_date) VALUES (?, ?)', customers_data)

orders_data = []
for cid in range(1, n_customers + 1):
    reg_date = reg_dates[cid - 1]
    n_orders = np.random.randint(0, 5)
    for _ in range(n_orders):
        days_after = np.random.randint(1, 90)
        ord_date = reg_date + pd.Timedelta(days=int(days_after))
        if ord_date > pd.Timestamp('2026-03-31'):
            ord_date = pd.Timestamp('2026-03-31')
        amount = round(np.random.uniform(50.0, 1000.0), 2)
        orders_data.append((cid, ord_date.strftime('%Y-%m-%d'), amount))

cursor.executemany('INSERT INTO cohort_orders (customer_id, order_date, amount) VALUES (?, ?, ?)', orders_data)
conn.commit()


query = '''
SELECT 
    strftime('%Y-%m', c.registration_date) AS cohort_month,
    CAST((julianday(o.order_date) - julianday(c.registration_date)) / 30 AS INTEGER) AS month_offset,
    COUNT(DISTINCT c.customer_id) AS unique_customers,
    COUNT(o.order_id) AS total_orders,
    SUM(o.amount) AS total_revenue,
    AVG(o.amount) AS avg_ltv
FROM cohort_customers c
LEFT JOIN cohort_orders o ON c.customer_id = o.customer_id
GROUP BY cohort_month, month_offset
'''

df_cohort = pd.read_sql_query(query, conn)
print("=== PODSTAWOWE DANE KOHORTOWE ===")
print(df_cohort.head(10))
print("\n" + "="*50 + "\n")



pivot_df = df_cohort.pivot(index='cohort_month', columns='month_offset', values='avg_ltv').fillna(0)


all_values = pivot_df.values.flatten()
all_values_nonzero = all_values[all_values > 0]

if len(all_values_nonzero) > 0:
    q25, q50, q75 = np.percentile(all_values_nonzero, [25, 50, 75])
else:
    q25, q50, q75 = 0, 0, 0


def map_to_symbol(val):
    if val == 0:
        return '.'    
    elif val <= q25:
        return '░'    
    elif val <= q50:
        return '▒'    
    elif val <= q75:
        return '▓'    
    else:
        return '█'    

heatmap_text = pivot_df.map(map_to_symbol)

print("=== TEKSTOWA HEATMAPA (LTV Kohort) ===")
print("Legenda: . (brak), ░ (niski), ▒ (średni), ▓ (wysoki), █ (bardzo wysoki)")
print("-" * 60)
print(heatmap_text)


conn.close()
```
