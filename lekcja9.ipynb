```python
# 1
class Produkt:
    def __init__(self,nazwa,cena,ilosc_w_magazyne):
        self.nazwa=nazwa
        self.cena=cena
        self.ilosc_w_magazynie=ilosc_w_magazyne

    def pokaz_info(self):
        print(f"Produkt: {self.nazwa} | Cena: {self.cena} zł | Ilość w magazynie: {self.ilosc_w_magazynie} szt.")

    def sprzedaj(self,ilosc):
        if ilosc <= self.ilosc_w_magazynie:
            self.ilosc_w_magazynie -= ilosc
            print(f"Sprzedano {ilosc} szt. Pozostało w magazynie: {self.ilosc_w_magazynie}")
        else:
            print("Brak wystarczającej ilości produktu w magazynie!")

    def dostawa(self,ilosc):
        self.ilosc_w_magazynie += ilosc
        print(f"Przyjęto dostawę {ilosc} szt. Aktualna ilość: {self.ilosc_w_magazynie}")


p1 = Produkt("Laptop", 3500.0, 10)
p1.pokaz_info()
p1.sprzedaj(3)
p1.dostawa(5)
p1.pokaz_info()
```

```python
# 11
# stworz klasy
# Uczen (imie, oceny=[])
# Klasa (nazwa, uczniowie=[])
# Metody Klasa :
# dodaj_ucznia(uczen)
# srednia_klasy() – średnia ocen wszystkich uczniów
# najlepszy_uczen() – zwraca ucznia z najwyższą średnią

class Uczen:
    def __init__(self, imie,oceny=None):
        self.imie=imie
        self.oceny=oceny if oceny is not None else []
    def srednia_ocen(self):
        if not self.oceny:
            return 0
        return sum(self.oceny)/len(self.oceny)

class Klasa:
    def __init__(self,nazwa):
        self.nazwa=nazwa
        self.uczniowie=[]
    def  dodaj_ucznia(self,uczen):
        self.uczniowie.append(uczen)

    def srednia_klasy(self):
        if not self.uczniowie:
            return 0
        srednie = [u.srednia_ocen() for u in self.uczniowie] 
        return sum(srednie) / len(srednie)

    def najlepszy_uczen(self):
        return max(self.uczniowie, key=lambda u: u.srednia_ocen())

u1 = Uczen("Anna", [5, 4, 5, 5])
u2 = Uczen("Jan", [3, 3, 4, 3])
u3 = Uczen("Kasia", [5, 5, 5, 4])

klasa_1a = Klasa("1A")
klasa_1a.dodaj_ucznia(u1)
klasa_1a.dodaj_ucznia(u2)
klasa_1a.dodaj_ucznia(u3)

print(klasa_1a)

print(f"Średnia całej klasy: {klasa_1a.srednia_klasy():.2f}")
najlepszy = klasa_1a.najlepszy_uczen()
print(f"Najlepszy uczeń: {najlepszy.imie} (średnia: {najlepszy.srednia_ocen():.2f})")
```

```python
# 12

# Stwórz klasę bazową Figura z metodą pole() .
# Stwórz klasy potomne:
# Prostokat(szerokosc, wysokosc)
# Kolo(promien)
# Każda klasa implementuje pole() .
# Stwórz listę figur i oblicz łączne pole.


class Figura:

    def pole(self):
        raise NotImplementedError(
            "Metoda pole() musi być zaimplementowana."
        )


class Prostokat(Figura):

    def __init__(self, szerokosc, wysokosc):
        self.szerokosc = szerokosc
        self.wysokosc = wysokosc

    def pole(self):
        return self.szerokosc * self.wysokosc
    
class Kolo(Figura):

    def __init__(self, promien):
        self.promien = promien

    def pole(self):
        return 3.14159 * (self.promien ** 2)


figury = [
    Prostokat(4, 5), 
    Kolo(3), 
    Prostokat(2, 3), 
    Kolo(2)
]

laczne_pole = sum(f.pole() for f in figury)

for nr, figura in enumerate(figury, 1):
    print(f"Figura {nr}: pole = {figura.pole():.2f}")

print(f"\nŁączne pole wszystkich figur: {laczne_pole:.2f}")
```

```python
# 15

# Klasy:
# Akcja (nazwa, cena_kupna, cena_aktualna)
# Portfel (akcje=[], kapital_poczatkowy)
# Metody Portfel :
# kup_akcje(akcja, ilosc)
# sprzedaj_akcje(akcja, ilosc)
# aktualizuj_ceny(slownik_cen) – aktualizuje cena_aktualna
# wartosc_portfela() – łączna wartość
# zysk_strata() – różnica względem kapitału początkowego
# Zasymuluj kupno/sprzedaż 5 akcji z różnymi zmianami cen.


class Akcja:

    def __init__(self, nazwa, cena_kupna):
        self.nazwa = nazwa
        self.cena_kupna = cena_kupna
        self.cena_aktualna = cena_kupna 


class Portfel:

    def __init__(self, kapital_poczatkowy):
        self.kapital_poczatkowy = kapital_poczatkowy
        self.gotowka = kapital_poczatkowy
        self.akcje = [] 

    def kup_akcje(self, akcja, ilosc):
        koszt = akcja.cena_kupna * ilosc
        if self.gotowka >= koszt:
            self.gotowka -= koszt
            # Sprawdzamy, czy akcja już jest w portfelu
            for pozycja in self.akcje:
                if pozycja["akcja"] == akcja:
                    pozycja["ilosc"] += ilosc
                    break
            else:
                self.akcje.append({"akcja": akcja, "ilosc": ilosc})
            print(f"Kupiono {ilosc} szt. akcji {akcja.nazwa} za {koszt} zł.")
        else:
            print(f"Brak środków na zakup {akcja.nazwa}!")

    def sprzedaj_akcje(self, akcja, ilosc):
        for pozycja in self.akcje:
            if pozycja["akcja"] == akcja:
                if pozycja["ilosc"] >= ilosc:
                    pozycja["ilosc"] -= ilosc
                    przychod = akcja.cena_aktualna * ilosc
                    self.gotowka += przychod
                    print(
                        f"Sprzedano {ilosc} szt. akcji {akcja.nazwa} za {przychod} zł."
                    )
                    if pozycja["ilosc"] == 0:
                        self.akcje.remove(pozycja)
                    return
                else:
                    print(f"Nie masz tylu akcji {akcja.nazwa} na sprzedaż!")
                    return
        print(f"Nie posiadasz akcji {akcja.nazwa} w portfelu!")

    def aktualizuj_ceny(self, slownik_cen):
        for pozycja in self.akcje:
            nazwa = pozycja["akcja"].nazwa
            if nazwa in slownik_cen:
                pozycja["akcja"].cena_aktualna = slownik_cen[nazwa]

    def wartosc_portfela(self):
        wartosc_akcji = sum(
            p["akcja"].cena_aktualna * p["ilosc"] for p in self.akcje
        )
        return self.gotowka + wartosc_akcji

    def zysk_strata(self):
        return self.wartosc_portfela() - self.kapital_poczatkowy


a1 = Akcja("AAPL", 180)
a2 = Akcja("TSLA", 240)
a3 = Akcja("MSFT", 410)
a4 = Akcja("GOOGL", 140)
a5 = Akcja("AMZN", 175)


moj_portfel = Portfel(10000)


moj_portfel.kup_akcje(a1, 10)  # koszt: 1800
moj_portfel.kup_akcje(a2, 5)  # koszt: 1200
moj_portfel.kup_akcje(a3, 4)  # koszt: 1640
moj_portfel.kup_akcje(a4, 15)  # koszt: 2100
moj_portfel.kup_akcje(a5, 10)  # koszt: 1750

print(f"Gotówka po zakupach: {moj_portfel.gotowka} zł\n")


nowe_ceny = {"AAPL": 195, "TSLA": 220, "MSFT": 430, "GOOGL": 135, "AMZN": 190}
moj_portfel.aktualizuj_ceny(nowe_ceny)


moj_portfel.sprzedaj_akcje(a2, 2)
moj_portfel.sprzedaj_akcje(a1, 5)


print("\n--- PODSUMOWANIE PORTFELA ---")
print(f"Całkowita wartość portfela: {moj_portfel.wartosc_portfela():.2f} zł")
zysk = moj_portfel.zysk_strata()
print(
    f"Wynik finansowy (Zysk/Strata): {zysk:+.2f} zł"
)
```
