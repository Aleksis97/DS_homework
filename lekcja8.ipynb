```python
# Napisz trzy funkcje: celsius_to_fahrenheit(c) , fahrenheit_to_celsius(f) , celsius_to_kelvin(c) .
# Dataset:
Temperatura_C = [0, 10, 20, 30, 37, 100]
# Wymagania:
# Każda funkcja zwraca skonwertowaną wartość
# Zastosuj funkcje do konwersji wszystkich temperatur (pętla for )
# Oczekiwany rezultat:
# Tabela z trzema kolumnami: °C, °F, K


def celsius_to_fahrenheit(c):
    return c * 1.8 + 32

def fahrenheit_to_celsius(f):
    return (f - 32) / 1.8

def celsius_to_kelvin(c):
    return c + 273.15

temperatura_c = [0, 10, 20, 30, 37, 100]

print(f"{'°C':<8} | {'°F':<8} | {'K':<8}")
print("-" * 28)

for c in temperatura_c:
    f = celsius_to_fahrenheit(c)
    k = celsius_to_kelvin(c)
    print(f"{c:<8} | {f:<8.1f} | {k:<8.2f}")
```

```python
# Napisz funkcję powitanie(imie, jezyk="polski") , która zwraca powitanie w wybranym języku (polski,
# angielski, niemiecki).
# Dataset:
# Brak (funkcja operuje na stringach)
# Wymagania:
# Domyślny język to polski
# Obsługa 3 języków przy pomocy if / elif / else
# Wywołaj funkcję 3 razy z różnymi parametrami
# Oczekiwany rezultat:
# powitanie("Anna") → "Cześć, Anna!"
# powitanie("Anna", "angielski") → "Hello, Anna!"
# powitanie("Anna", "niemiecki") → "Hallo, Anna!"

def powitanie(imie, jezyk="polski"):
    if jezyk=="polski":  
        return(f"Cześć, {imie}!")
    elif jezyk=="angielski":  
        return(f"Hello, {imie}!") 
    else: return(f"Hallo, {imie}!") 


print(powitanie("Anna"))                  
print(powitanie("Anna", "angielski"))     
print(powitanie("Anna", "niemiecki"))     
```

```python
# Napisz funkcję normalize_minmax(dane) , która normalizuje listę liczb do zakresu [0, 1] używając
# metody min-max.
# Wymagania:
# Wzór: (x - min) / (max - min)
# Funkcja zwraca nową listę znormalizowanych wartości
# Zaokrąglij do 3 miejsc po przecinku
# Rozwiązanie w dwóch wariantach: (a) pętla for , (b) list comprehension
# Oczekiwany rezultat:
# Lista wartości w zakresie [0.0, 1.0]
# Minimum = 0.0, Maximum = 1.0


def normalize_minmax(dane,metoda="min-max",zaokraglenie=3):
    if metoda=='min-max':
        min_val = min(dane)
        max_val = max(dane)
        zakres = max_val - min_val

        znormalizowane = []

        for x in dane:
            norma=(x - min_val) / zakres
            znormalizowane.append(round(norma,zaokraglenie))
        return znormalizowane

        
wynagrodzenia = [3500, 4200, 5800, 7200, 9500, 4800, 6100]
normalize_minmax(wynagrodzenia)
print(normalize_minmax(wynagrodzenia))
```

```python
def normalize_minmax_lc(dane):
    min_val = min(dane)
    zakres = max(dane) - min_val
    
    return [round((x - min_val) / zakres, 3) for x in dane]

wynagrodzenia = [3500, 4200, 5800, 7200, 9500, 4800, 6100]
print(normalize_minmax_lc(wynagrodzenia))
```

```python
klienci = [
{"id": 1, "wiek": 25, "dochod": 4500, "wydatki": 3200},
{"id": 2, "wiek": 200, "dochod": 7800, "wydatki": 5100}, # outlier wiek
{"id": 3, "wiek": 45, "dochod": 9200, "wydatki": -500}, # outlier wydatki
{"id": 4, "wiek": 30, "dochod": 6000, "wydatki": 4000},
{"id": 5, "wiek": 50, "dochod": 11000, "wydatki": 6500},
{"id": 6, "wiek": 28, "dochod": 5000, "wydatki": 3500},
]
# Stwórz kompletny pipeline analizy danych klientów: czyszczenie outlierów → normalizacja → segmentacja.
# Wymagania:
# Funkcja usun_outliery(klienci) - usuwa rekordy z nieprawidłowym wiekiem ( <0 lub >120 ) lub ujemnymi wydatkami
# Funkcja normalizuj_pole(klienci, pole) - normalizacja min-max dla wskazanego klucza, zwraca nową listę z dodatkowym polem <pole>_norm
# Funkcja segmentuj(klienci) - przypisuje segment niski / sredni / wysoki według dochod
# Połącz wszystko używając list comprehensions i lambda ( sorted z lambda do wypisania segmentów po
# dochodzie)
# Oczekiwany rezultat:
# Czyste dane z segmentami i znormalizowanymi wartościami wieku i dochodu

def usun_outliery(dane):
    return [k for k in dane if 0 <= k["wiek"] <= 120 and k["wydatki"] >= 0]

def normalizuj_pole(dane, pole):
    wartosci = [k[pole] for k in dane]
    # print(F"{ k},{k["wiek"]}")
    min_val, max_val = min(wartosci), max(wartosci)
    zakres = max_val - min_val
    
    if zakres == 0:
        return [{**k, f"{pole}_norm": 0.0} for k in dane]
        
    return [
        {**k, f"{pole}_norm": round((k[pole] - min_val) / zakres, 3)}
        for k in dane
    ]

def segmentuj(dane):
    return [
        {
            **k,
            "segment": "niski" if k["dochod"] < 5000 else ("sredni" if k["dochod"] <= 10000 else "wysoki")
        }
        for k in dane
    ]


pipeline = segmentuj(
    normalizuj_pole(
        normalizuj_pole(
            usun_outliery(klienci), 
            "wiek"
        ), 
        "dochod"
    )
)

# Sortowanie wyników według dochodu rosnąco przy użyciu sorted() i lambda
finalni_klienci = sorted(pipeline, key=lambda k: k["dochod"])

for k in finalni_klienci:
    print(k)
```

```python
# 13
sprzedaz = [
    (1, "elektronika", 45, 12500),
    (1, "odziez", 78, 4200),
    (2, "elektronika", 52, 15200),
    (2, "odziez", 65, 3800),
    (3, "elektronika", 38, 9800),
    (3, "odziez", 82, 5100),
    (4, "elektronika", 61, 18900),
    (4, "odziez", 71, 4500),
]

def tuple_to_dict(dane):
    return [
        {
            "dzien": item[0],
            "kategoria": item[1],
            "liczba_transakcji": item[2],
            "wartosc": item[3]
        }
        for item in dane
    ]

def dodaj_srednia_transakcji(dane):
    return [
        {
            **k,
            "srednia_transakcji": round(k["wartosc"] / k["liczba_transakcji"], 2)
        }
        for k in dane
    ]

def agreguj_po_kategorii(dane):
    agregaty = {}
    for k in dane:
        kat = k["kategoria"]
        agregaty[kat] = agregaty.get(kat, 0) + k["wartosc"]
    return agregaty


krok_1 = tuple_to_dict(sprzedaz)
krok_2 = dodaj_srednia_transakcji(krok_1)
krok_3 = agreguj_po_kategorii(krok_2)

# Sortowanie kategorii według wartości malejąco używając sorted() i lambda
posortowane = sorted(krok_3.items(), key=lambda x: x[1], reverse=True)

for kategoria, suma in posortowane:
    print(f"Kategoria: {kategoria} | Łączna wartość: {suma}")
```
