```python
import numpy as np


v1 = np.array([3, 4])
v2 = np.array([1, 2])


suma = v1 + v2
diff = v1 - v2

wektory = {
    "v1": v1,
    "v2": v2,
    "suma": suma,
    "diff": diff
}

# L2
normy = {label: np.linalg.norm(v) for label, v in wektory.items()}
# print(normy)

def narysuj_wektory_2d(v1, v2, suma, diff, labels):
    print("--- ASCII Siatka wektorów ---")
    print(f"{labels[0]}: {v1}")
    print(f"{labels[1]}: {v2}")
    print(f"{labels[2]}: {suma}")
    print(f"{labels[3]}: {diff}")
    print("-----------------------------")

narysuj_wektory_2d(v1, v2, suma, diff, labels=["v1", "v2", "v1+v2", "v1-v2"])

# Wypisanie tabeli
print(f"{'Label':<10} | {'Współrzędne':<12} | {'Długość L2':<10}")
print("-" * 38)
for label, v in wektory.items():
    print(f"{label:<10} | {str(v):<12} | {normy[label]:<10.4f}")

```

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

iloczyn_skalarny = np.dot(a, b)

norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

cos_theta = iloczyn_skalarny / (norm_a * norm_b)
kat_radianach = np.arccos(cos_theta)
kat_stopnie = np.degrees(kat_radianach)

print(f"Iloczyn skalarny: {iloczyn_skalarny}")
print(f"Kąt między wektorami: {kat_stopnie:.2f}°")
```

```python
import numpy as np

def make_synthetic_flowers(seed=42):
    rng = np.random.default_rng(seed)
    X = rng.uniform(low=1.0, high=7.0, size=(100, 4))
    # print(X)
    y = rng.integers(0, 3, size=100)
    feature_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
    target_names = ['setosa', 'versicolor', 'virginica']
    return X, y, feature_names, target_names

X, y, feature_names, target_names = make_synthetic_flowers(seed=42)

#  rowvar=False- kolumny to cechy
cov_matrix = np.cov(X, rowvar=False)
corr_matrix = np.corrcoef(X, rowvar=False)

# 3. Funkcja pomocnicza do ładnego wyświetlania tabeli macierzy
def print_matrix_table(matrix, title, names):
    print(f"\n--- {title} ---")
    header = f"{'':<15}" + "".join([f"{name:<15}" for name in names])
    print(header)
    print("-" * len(header))
    for i, row in enumerate(matrix):
        row_str = f"{names[i]:<15}" + "".join([f"{val:<15.4f}" for val in row])
        print(row_str)

print_matrix_table(cov_matrix, "Macierz Kowariancji", feature_names)
print_matrix_table(corr_matrix, "Macierz Korelacji", feature_names)

# pary cech o najwyższej korelacji 
masked_corr = np.abs(corr_matrix)
np.fill_diagonal(masked_corr, 0)
# print(masked_corr)
max_idx = np.argmax(masked_corr)
# print(max_idx)
row, col = np.unravel_index(max_idx, corr_matrix.shape)
print(np.unravel_index(max_idx, corr_matrix.shape))
(np.int64(2), np.int64(3))

feature_1 = feature_names[row]
feature_2 = feature_names[col]
max_corr_val = corr_matrix[row, col]

print(f"\nNajsilniejsza korelacja:")
print(f"Para cech: {feature_1} oraz {feature_2}")
print(f"Wartość współczynnika korelacji: {max_corr_val:.4f}")


print(f"\nPierwsze 10 wartości dla pary ({feature_1}, {feature_2}):")
print(f"{feature_1:<15} | {feature_2:<15}")
print("-" * 33)
for i in range(10):
    print(f"{X[i, row]:<15.4f} | {X[i, col]:<15.4f}")



if abs(max_corr_val) > 0.7:
    strength = "bardzo silna"
elif abs(max_corr_val) > 0.4:
    strength = "umiarkowana"
else:
    strength = "słaba"

direction = "dodatnia" if max_corr_val > 0 else "ujemna"
print(f"Zależność między cechami {feature_1} i {feature_2} ma charakter **{strength}** i jest **{direction}**.")
```

```python
import numpy as np
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df = df.dropna(subset=['Fare']).copy()
fare = df['Fare'].to_numpy()
pclass = df['Pclass'].to_numpy()

group_means = np.empty_like(fare, dtype=float)
group_stds = np.empty_like(fare, dtype=float)


for c in [1, 2, 3]:
    mask = (pclass == c)
    c_fares = fare[mask]

    group_means[mask] = c_fares.mean()
    group_stds[mask] = c_fares.std(ddof=0)  


fare_normalized_np = (fare - group_means) / group_stds


df['Fare_normalized_pd'] = df.groupby("Pclass")["Fare"].transform(
    lambda x: (x - x.mean()) / x.std(ddof=0)
)
fare_normalized_pd = df['Fare_normalized_pd'].to_numpy()


identyczne = np.allclose(fare_normalized_np, fare_normalized_pd, atol=1e-9)
print(f"Czy wyniki Ścieżki A i Ścieżki B są identyczne? {identyczne}\n")
df['Fare_normalized_np'] = fare_normalized_np

stats_table = df.groupby('Pclass')['Fare_normalized_np'].agg(
    Mean=lambda x: round(x.mean(), 10),
    Std=lambda x: round(x.std(ddof=0), 10)
).reset_index()

print("--- Weryfikacja statystyk per klasa (Pclass) ---")
print(stats_table.to_string(index=False))
```
