```python
import pandas as pd
import numpy as np
# Wczytujemy dane Titanic
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)

sredni_wiek_klasa_plec = titanic.groupby(['Pclass', 'Sex'])['Age'].mean().reset_index()
sredni_wiek_klasa_plec.columns = ['Klasa (Pclass)', 'Płeć (Sex)', 'Średni Wiek']
print("\n=== Średni wiek według klasy ===")
print(sredni_wiek_klasa_plec)
print(sredni_wiek_klasa_plec.to_string(index=False))

```

```python
# Dla datasetu Titanic policz, ile pasażerów było w każdej klasie i czy przeżyli ( Survived ).
# Dataset: Titanic
# Wymagania:
# Grupuj według Pclass i Survived
# Użyj .size() lub .count()
# Przedstaw wyniki jako tabelę pivot ( pivot_table lub unstack() )
# Oczekiwany rezultat:
# Tabela pokazująca liczbę pasażerów dla każdej kombinacji klasy i statusu przeżycia


sredni_wiek_klasa_plec = titanic.groupby(['Pclass','Survived'])['PassengerId'].count().reset_index()
print(sredni_wiek_klasa_plec)

print('------------------------')
survival_pivot = titanic.groupby(['Pclass', 'Survived'])['PassengerId'].count().unstack() 
print(survival_pivot)


```

```python
# Dla syntetycznego datasetu sprzedaży oblicz dla każdego regionu: minimum, maksimum, średnią i
# odchylenie standardowe sprzedaży.
import numpy as np
import pandas as pd
np.random.seed(42)
sales = pd.DataFrame({
'Region': np.random.choice(['A', 'B', 'C'], 100),
'Sprzedaż': np.random.randint(50, 500, 100)
})
# Wymagania:
# Użyj .agg(['min', 'max', 'mean', 'std'])
# Zaokrąglij wyniki do 2 miejsc po przecinku


survival_grouped = sales.groupby(['Region'])['Sprzedaż'].agg(['min','max', 'mean', 'std'])

print(survival_grouped)
print(survival_grouped.round(2))
```

```python
# Dodaj do datasetu Titanic nową kolumnę Avg_Fare_Pclass , która zawiera średnią opłatę ( Fare ) dla każdej
# klasy.
# Dataset: Titanic
# Wymagania:
# Użyj .groupby().transform('mean')
# Dodaj kolumnę do oryginalnego DataFrame
# Wyświetl pierwsze 10 wierszy z kolumnami: Pclass , Fare , Avg_Fare_Pclass
# Oczekiwany rezultat:
# DataFrame z nową kolumną — każdy wiersz ma średnią swojej klasy



url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)


titanic['Avg_Fare_Pclass'] = titanic.groupby('Fare')['Pclass'].transform('mean')


print(titanic[['Pclass', 'Fare', 'Avg_Fare_Pclass']].head(10))
```

```python
# Podziel wiek pasażerów Titanica na 4 kategorie: Dziecko (0-12), Nastolatek (13-18), Dorosły (19-60), Senior(60+).
# Dataset: Titanic
# Wymagania:
# Użyj pd.cut() z własnymi przedziałami
# Oblicz survival rate dla każdej kategorii wiekowej
# Przedstaw wyniki jako tekstowy słupek ( █ = 2%)

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)


titanic['Age_Group'] = pd.cut(
titanic['Age'],
bins=[-1, 12, 18, 60, 120],
labels=[' Dziecko', 'Nastolatek', 'Dorosły', 'Senior']
)


survival_age = titanic.groupby('Age_Group', observed=False)['Survived'].agg(['count',
'mean'])
# print(survival_age)
survival_age.columns = ['Liczba', 'Survival_Rate']
print("\n=== Survival rate według grup wiekowych ===")
print(survival_age.round(3))



print('====')
for k, v in survival_age['Survival_Rate'].items():
    print(k, v)

print("\nTekstowe porównanie (█ = 2%):")
for age_group, rate in survival_age['Survival_Rate'].items():
    pct = rate * 100
    bar = '█' * int(pct // 2)
    print(f" {str(age_group):10s} |{bar} {pct:.1f}%")
```

```python
# Stwórz nową kolumnę Family_Size w Titanic jako sumę SibSp + Parch + 1 (sam pasażer).
# Dataset: Titanic
# Wymagania:
# Oblicz Family_Size
# Oblicz survival rate według rozmiaru rodziny
# Zidentyfikuj, który rozmiar rodziny miał najwyższy survival rate
# Narysuj tekstowe słupki porównujące wszystkie rozmiary


url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)


titanic['Family_Size'] = titanic['SibSp'] + titanic['Parch'] + 1


family_survival = titanic.groupby('Family_Size')['Survived'].agg(['count', 'mean'])
family_survival.columns = ['Liczba', 'Survival_Rate']
family_survival['Survival_Pct'] = family_survival['Survival_Rate'] * 100

# 3. Identyfikacja rozmiaru rodziny z najwyższym survival rate
best_size = family_survival['Survival_Rate'].idxmax()
best_rate = family_survival.loc[best_size, 'Survival_Pct']

print(f"Najwyższy survival rate ({best_rate:.1f}%) odnotowano dla rozmiaru rodziny: {best_size}\n")

# 4. Tekstowy wykres słupkowy (█ = 2%)
print("Tekstowe porównanie survival rate (█ = 2%):")
for size, row in family_survival.iterrows():
    bar = '█' * int(row['Survival_Pct'] // 2)
    print(f" Rozmiar {size:2d} |{bar} {row['Survival_Pct']:.1f}% (liczba osób: {int(row['Liczba'])})")

```
