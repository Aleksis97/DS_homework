```python
# Podziel wiek pasażerów Titanica na 4 kategorie: Dziecko (0-12), Nastolatek (13-18), Dorosły (19-60), Senior(60+).
# Dataset: Titanic
# Wymagania:
# Użyj pd.cut() z własnymi przedziałami
# Oblicz survival rate dla każdej kategorii wiekowej
# Przedstaw wyniki jako tekstowy słupek ( █ = 2%)

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)


titanic['Age_Group'] = pd.cut(
titanic['Age'],
bins=[-1, 12, 18, 60, 120],
labels=[' Dziecko', 'Nastolatek', 'Dorosły', 'Senior']
)


survival_age = titanic.groupby('Age_Group', observed=False)['Survived'].agg(['count',
'mean'])
# print(survival_age)
survival_age.columns = ['Liczba', 'Survival_Rate']
print("\n=== Survival rate według grup wiekowych ===")
print(survival_age.round(3))



print('====')
for k, v in survival_age['Survival_Rate'].items():
    print(k, v)

print("\nTekstowe porównanie (█ = 2%):")
for age_group, rate in survival_age['Survival_Rate'].items():
    pct = rate * 100
    bar = '█' * int(pct // 2)
    print(f" {str(age_group):10s} |{bar} {pct:.1f}%")
```

```python
# Stwórz nową kolumnę Family_Size w Titanic jako sumę SibSp + Parch + 1 (sam pasażer).
# Dataset: Titanic
# Wymagania:
# Oblicz Family_Size
# Oblicz survival rate według rozmiaru rodziny
# Zidentyfikuj, który rozmiar rodziny miał najwyższy survival rate
# Narysuj tekstowe słupki porównujące wszystkie rozmiary


url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)


titanic['Family_Size'] = titanic['SibSp'] + titanic['Parch'] + 1


family_survival = titanic.groupby('Family_Size')['Survived'].agg(['count', 'mean'])
family_survival.columns = ['Liczba', 'Survival_Rate']
family_survival['Survival_Pct'] = family_survival['Survival_Rate'] * 100


best_size = family_survival.sort_values(by='Survival_Rate', ascending=False).index[0]
best_rate = family_survival.loc[best_size, 'Survival_Pct']

print(family_survival)
print(f"Najwyższy survival rate ({best_rate:.1f}%) odnotowano dla rozmiaru rodziny: {best_size}\n")


print("Tekstowe porównanie survival rate (█ = 2%):")
for size, row in family_survival.iterrows():
    bar = '█' * int(row['Survival_Pct'] // 2)
    print(f" Rozmiar {size:2d} |{bar} {row['Survival_Pct']:.1f}% (liczba osób: {int(row['Liczba'])})")

```

```python
# Dla syntetycznego datasetu California Housing (funkcja make_california_housing z lekcji) znajdź 5
# regionów z najwyższą średnią ceną domu.
# Wymagania:
# Wygeneruj dataset: df_housing = make_california_housing() (skopiuj funkcję z sekcji „California
# Housing")
# Stwórz regiony na podstawie Latitude / Longitude — podziel Kalifornię na siatkę 4×4 (16 regionów),
# np. df['lat_bin'] = pd.cut(df['Latitude'], 4) i analogicznie dla długości; łącz w etykietę region
# = str(lat_bin) + '|' + str(lon_bin)
# Oblicz średnią cenę ( MedHouseVal ) według regionu
# Posortuj malejąco i wyświetl top 5

import pandas as pd
import numpy as np

# === GENEROWANIE SYNTETYCZNEGO DATASETU CALIFORNIA HOUSING ===
def make_california_housing(n=5000, seed=42):
    """Generuje syntetyczny dataset wzorowany na California Housing."""
    rng = np.random.default_rng(seed)
    
    # Współrzędne geograficzne w obszarze Kalifornii
    latitude = rng.uniform(32.5, 41.5, n)
    longitude = rng.uniform(-123.5, -114.5, n)
    
    # Dochód: rozkład skośny
    med_inc = rng.gamma(shape=2.0, scale=1.8, size=n) + 0.5
    
    # Wiek domu
    house_age = rng.integers(1, 55, n)
    
    # Pokoje/sypialnie/populacja
    ave_rooms = rng.normal(5.5, 1.5, n).clip(1, 15)
    ave_bedrms = (ave_rooms * rng.uniform(0.15, 0.25, n)).clip(0.5, 5)
    population = rng.integers(50, 5000, n)
    ave_occup = rng.normal(3.0, 0.8, n).clip(1, 10)
    
    # Cena domu (w 100_000 USD) - zależna od dochodu, wieku i lokalizacji
    med_house_val = (
        med_inc * 0.6
        + (latitude - 36) * 0.15
        - (house_age / 100)
        + rng.normal(0, 0.3, n)
    ).clip(0.15, 5.0)
    
    return pd.DataFrame({
        'MedInc': med_inc.round(4),
        'HouseAge': house_age,
        'AveRooms': ave_rooms.round(4),
        'AveBedrms': ave_bedrms.round(4),
        'Population': population,
        'AveOccup': ave_occup.round(4),
        'Latitude': latitude.round(4),
        'Longitude': longitude.round(4),
        'MedHouseVal': med_house_val.round(4)
    })

df_housing = make_california_housing(n=5000, seed=42)
# print("=== Syntetyczny California Housing Dataset ===")
# print(df_housing.head())
# print(f"\nShape: {df_housing.shape}")
# print(f"\nOpis:\n{df_housing.describe().round(3)}")




df_housing['lat_bin'] = pd.cut(df_housing['Latitude'], bins=4)
df_housing['lon_bin'] = pd.cut(df_housing['Longitude'], bins=4)


df_housing['region'] = df_housing['lat_bin'].astype(str) + ' | ' + df_housing['lon_bin'].astype(str)


top_regions = (
    df_housing.groupby('region')['MedHouseVal']
    .mean()
    .reset_index()
    .sort_values(by='MedHouseVal', ascending=False)
    .head(5)
)

print("=== Top 5 regionów z najwyższą średnią ceną domów ===")
print(top_regions.to_string(index=False))
```

```python

# Wymagania:
# Stwórz kolumnę Is_Alone (czy SibSp + Parch == 0 )
# Grupuj według 3 kolumn: Pclass , Sex , Is_Alone
# Oblicz średnią wieku i survival rate
# Spłaszcz MultiIndex przez .reset_index()
# Oczekiwany rezultat:
# Czytelna tabela z wynikami dla wszystkich kombinacji

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)


titanic['Is_Alone'] = (titanic['SibSp'] + titanic['Parch'] == 0).astype(int)

# 2. Grupowanie według Pclass, Sex i Is_Alone oraz obliczenie średnich
result = (
    titanic.groupby(['Pclass', 'Sex', 'Is_Alone'])
    .agg(
        Mean_Age=('Age', 'mean'),
        Survival_Rate=('Survived', 'mean')
    )
    .reset_index()
)

# 3. Wyświetlenie tabeli z wynikami

print(result.to_string(index=False))
```

```python
import pandas as pd
import numpy as np


def make_california_housing(n=5000, seed=42):
    rng = np.random.default_rng(seed)
    
    latitude = rng.uniform(32.5, 41.5, n)
    longitude = rng.uniform(-123.5, -114.5, n)
    
    med_inc = rng.gamma(shape=2.0, scale=1.8, size=n) + 0.5
    house_age = rng.integers(1, 55, n)
    ave_rooms = rng.normal(5.5, 1.5, n).clip(1, 15)
    ave_bedrms = (ave_rooms * rng.uniform(0.15, 0.25, n)).clip(0.5, 5)
    population = rng.integers(50, 5000, n)
    ave_occup = rng.normal(3.0, 0.8, n).clip(1, 10)
    
    med_house_val = (
        med_inc * 0.6
        + (latitude - 36) * 0.15
        - (house_age / 100)
        + rng.normal(0, 0.3, n)
    ).clip(0.15, 5.0)
    
    return pd.DataFrame({
        'MedInc': med_inc.round(4),
        'HouseAge': house_age,
        'AveRooms': ave_rooms.round(4),
        'AveBedrms': ave_bedrms.round(4),
        'Population': population,
        'AveOccup': ave_occup.round(4),
        'Latitude': latitude.round(4),
        'Longitude': longitude.round(4),
        'MedHouseVal': med_house_val.round(4)
    })

df_housing = make_california_housing(n=5000, seed=42)

df_housing['lat_bin'] = pd.cut(df_housing['Latitude'], bins=4)
df_housing['lon_bin'] = pd.cut(df_housing['Longitude'], bins=4)
df_housing['Region'] = df_housing['lat_bin'].astype(str) + ' | ' + df_housing['lon_bin'].astype(str)


def analyze_outliers_and_box(group):
    q1 = group.quantile(0.25)
    q3 = group.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    

    outliers_count = ((group < lower_bound) | (group > upper_bound)).sum()
    
    return pd.Series({
        'Outliers_Count': outliers_count,
        'Min': group.min(),
        'Q1': q1.round(4),
        'Median': group.median().round(4),
        'Q3': q3.round(4),
        'Max': group.max()
    })


regional_analysis = df_housing.groupby('Region')['MedHouseVal'].apply(analyze_outliers_and_box).unstack()

print("=== Tabela liczby outlierów oraz tekstowy boxplot (podsumowanie 5-punktowe) według regionu ===")
print(regional_analysis.to_string())
```

```python
import pandas as pd
import numpy as np


def make_california_housing(n=5000, seed=42):
    rng = np.random.default_rng(seed)
    
    latitude = rng.uniform(32.5, 41.5, n)
    longitude = rng.uniform(-123.5, -114.5, n)
    
    med_inc = rng.gamma(shape=2.0, scale=1.8, size=n) + 0.5
    house_age = rng.integers(1, 55, n)
    ave_rooms = rng.normal(5.5, 1.5, n).clip(1, 15)
    ave_bedrms = (ave_rooms * rng.uniform(0.15, 0.25, n)).clip(0.5, 5)
    population = rng.integers(50, 5000, n)
    ave_occup = rng.normal(3.0, 0.8, n).clip(1, 10)
    
    med_house_val = (
        med_inc * 0.6
        + (latitude - 36) * 0.15
        - (house_age / 100)
        + rng.normal(0, 0.3, n)
    ).clip(0.15, 5.0)
    
    return pd.DataFrame({
        'MedInc': med_inc.round(4),
        'HouseAge': house_age,
        'AveRooms': ave_rooms.round(4),
        'AveBedrms': ave_bedrms.round(4),
        'Population': population,
        'AveOccup': ave_occup.round(4),
        'Latitude': latitude.round(4),
        'Longitude': longitude.round(4),
        'MedHouseVal': med_house_val.round(4)
    })

df_housing = make_california_housing(n=5000, seed=42)


df_housing['lat_bin'] = pd.cut(df_housing['Latitude'], bins=10)
df_housing['lon_bin'] = pd.cut(df_housing['Longitude'], bins=10)
df_housing['Region'] = df_housing['lat_bin'].astype(str) + ' | ' + df_housing['lon_bin'].astype(str)


region_stats = df_housing.groupby('Region').agg(
    Mean_Price=('MedHouseVal', 'mean'),
    Median_Income=('MedInc', 'median'),
    Total_Population=('Population', 'sum')
).reset_index()

# === TOP 10 NAJDROŻSZYCH I NAJTAŃSZYCH REGIONÓW ===
top_10 = region_stats.sort_values(by='Mean_Price', ascending=False).head(10)
bottom_10 = region_stats.sort_values(by='Mean_Price', ascending=True).head(10)

print("=== Top 10 najdroższych regionów ===")
print(top_10.to_string(index=False))

print("\n=== Top 10 najtańszych regionów ===")
print(bottom_10.to_string(index=False))



counts, xedges, yedges = np.histogram2d(
    df_housing['Longitude'], df_housing['Latitude'], bins=10
)
sum_vals, _, _ = np.histogram2d(
    df_housing['Longitude'], df_housing['Latitude'], bins=10, weights=df_housing['MedHouseVal']
)


mean_grid = np.divide(sum_vals, counts, out=np.zeros_like(sum_vals), where=counts!=0)
mean_grid_flat = mean_grid[mean_grid > 0]


q25, q50, q75 = np.percentile(mean_grid_flat, [25, 50, 75])


char_map = np.empty(mean_grid.shape, dtype=str)
for i in range(mean_grid.shape[0]):
    for j in range(mean_grid.shape[1]):
        val = mean_grid[i, j]
        if counts[i, j] == 0:
            char_map[i, j] = ' '
        elif val <= q25:
            char_map[i, j] = '·'
        elif val <= q50:
            char_map[i, j] = '▃'
        elif val <= q75:
            char_map[i, j] = '▇'
        else:
            char_map[i, j] = '█'

print("\n=== Tekstowa heatmapa 10x10 (MedHouseVal) ===")
for row in reversed(char_map.T):
    print(" ".join(row))
```
