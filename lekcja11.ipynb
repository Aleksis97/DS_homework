```python
# Wczytaj dataset Titanic w batchach po 100 wierszy i oblicz średni wiek w każdym batchu.
# Wymagania:
# Użyj chunksize=100
# Dla każdego batcha wypisz numer i średni wiek
# Oblicz globalną średnią ze wszystkich batchy (akumuluj age_sum i age_count )
import pandas as pd

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
# Wczytywanie w batchach (po 100 wierszy)
batch_size = 100
chunks = []
print("Wczytywanie w batchach:")
for i, chunk in enumerate(pd.read_csv(url, chunksize=batch_size)):
    print(f"Batch {i+1}: {len(chunk)} wierszy, średni wiek: {chunk['Age'].mean():.2f}")
    chunks.append(chunk)


#     if i >= 2: # Dla przykładu, tylko 3 batche
#         break
# # Połączenie wszystkich batchy (jeśli potrzebne)
# df_combined = pd.concat(chunks, ignore_index=True)
# print(f"\nPołączono {len(chunks)} batchy: {len(df_combined)} wierszy total")




print("\nPrzetwarzanie bez ładowania całości:")
total_rows = 0
age_sum = 0
age_count = 0
for chunk in pd.read_csv(url, chunksize=100):
    total_rows += len(chunk)
    age_sum += chunk['Age'].sum()
    age_count += chunk['Age'].count()
print(f"Przetworzono {total_rows} wierszy")
print(f"Średni wiek (wszystkie dane): {age_sum / age_count:.2f}")
```

```python
# Wyczyść dataset Titanic, stosując pełny pipeline:
# Usuń kolumny z >60% NaN
# Wypełnij Age medianą
# Wypełnij Embarked dominantą
# Usuń wiersze z pozostałymi NaN
# Wymagania:
# Przed i po: wypisz kształt DataFrame
# Sprawdź, czy nie ma NaN
# Porównaj statystyki Age ( describe() ) przed i po


titanic=pd.read_csv(url)
print(titanic.describe())
print("Kształt przed:", titanic.shape)

# rows_with_nan = titanic.isnull().any(axis=0)
# print(rows_with_nan)
# procent_nan2 = (titanic.isnull().mean()) * 100
# print(procent_nan2)

titanic=titanic.loc[:, titanic.isnull().mean() <= 0.6]

titanic['Age']=titanic['Age'].fillna(titanic['Age'].median())
embarked_mode = titanic['Embarked'].mode()[0]
print(embarked_mode)
titanic['Embarked']= titanic['Embarked'].fillna(embarked_mode)       
titanic = titanic.dropna()
print("Liczba pozostałych NaN:", titanic.isnull().sum().sum())
print("Kształt po:", titanic.shape)
print(titanic['Age'].describe())


```

```python
# Stwórz nowe zmienne z Titanic:
# FamilySize = SibSp + Parch + 1
# IsAlone = 1 jeśli FamilySize == 1 , inaczej 0
# AgeGroup = kategorie (0-12, 13-18, 19-35, 36-60, 60+) — użyj pd.cut
# Wymagania:
# Dodaj wszystkie 3 zmienne do DataFrame
# Wyświetl value_counts() dla każdej
# Oblicz wskaźnik przeżycia dla IsAlone == 1 vs IsAlone == 0



titanic=pd.read_csv(url)
# print(titanic)


pd.set_option('display.max_columns', None)
print(titanic.columns)

titanic['FamilySize']=titanic['SibSp']+titanic['Parch'] +1


titanic['IsAlone']=(titanic['FamilySize'] == 1).astype(int)

# import numpy as np
# titanic['IsAlone'] = np.where(titanic['FamilySize'] == 1, 1, 0)

bins = [-1, 12, 18, 35, 60, float('inf')]
labels = ['0-12', '13-18', '19-35', '36-60', '60+']
titanic['AgeGroup'] = pd.cut(titanic['Age'], bins=bins, labels=labels)

print(titanic.columns)


print(titanic['FamilySize'].value_counts())

print(titanic['IsAlone'].value_counts())

print(titanic['AgeGroup'].value_counts())


# print(titanic.query("IsAlone == 1").value_counts())
# print(titanic.query("IsAlone == 0").value_counts())


survival_by_alone = titanic.groupby('IsAlone')['Survived'].mean()
print(survival_by_alone)
```

```python
# Zoptymalizuj typy danych w Titanicu, aby zużywał mniej pamięci.
# Wymagania:
# Zmień Pclass , Survived , SibSp , Parch na int8
# Zmień Sex , Embarked na category
# Przed i po: wypisz zużycie pamięci ( .memory_usage(deep=True).sum() )
# Oblicz % redukcj

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

import pandas as pd
titanic=pd.read_csv(url)
mem_before = titanic.memory_usage(deep=True).sum() / 1024
print(mem_before)
titanic['Pclass'] = titanic['Pclass'].astype('int8')
titanic['Survived'] = titanic['Survived'].astype('int8')
titanic['SibSp'] = titanic['SibSp'].astype('int8')
titanic['Parch'] = titanic['Parch'].astype('int8')

titanic['Sex'] = titanic['Sex'].astype('category')
titanic['Embarked'] = titanic['Embarked'].astype('category')
mem_after = titanic.memory_usage(deep=True).sum() / 1024
print(mem_after)
reduction = (1 - mem_after / mem_before) * 100
print(f"{reduction:.2%}")
```

```python
# Wczytaj Titanic w batchach i oblicz agregacje bez ładowania całości do pamięci.
# Wymagania:
# chunksize=50
# Oblicz: total_passengers , avg_age , survival_rate — bez konkatenacji wszystkich batchy
# Użyj zmiennych akumulacyjnych ( sum , count , survived_sum )
# Porównaj wynik z ładowaniem pełnego DataFrame
# Oczekiwany rezultat:
# Te same statystyki, przy znacznie mniejszym szczytowym zużyciu pamięci

import pandas as pd
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic=pd.read_csv(url)


import pandas as pd

# Inicjalizacja zmiennych akumulacyjnych
total_passengers = 0
age_sum = 0
age_count = 0
survived_sum = 0

# Wczytywanie danych w batchach (chunkach)
for chunk in pd.read_csv(url, chunksize=50):
   total_passengers += len(chunk)

  # Agregacja wieku (pomijamy wartości NaN)
   valid_ages = chunk['Age'].dropna()
   age_sum += valid_ages.sum()
   age_count += len(valid_ages)

  # Agregacja przeżyć (pomijamy wartości NaN, jeśli występują)
   valid_survived = chunk['Survived'].dropna()
   survived_sum += valid_survived.sum()

# Obliczenie ostatecznych statystyk z batchy
batch_avg_age = age_sum / age_count if age_count > 0 else 0
batch_survival_rate = (
    survived_sum / total_passengers if total_passengers > 0 else 0
)

print('--- Wynik przetwarzania wsadowego (Batches) ---')
print(f'Liczba pasażerów: {total_passengers}')
print(f'Średni wiek: {batch_avg_age:.2f}')
print(f'Wskaźnik przeżycia: {batch_survival_rate:.4f}')

# Porównanie z pełnym DataFrame
df_full = pd.read_csv(url)
full_total = len(df_full)
full_avg_age = df_full['Age'].mean()
full_survival_rate = df_full['Survived'].mean()

print('\n--- Wynik z pełnego DataFrame ---')
print(f'Liczba pasażerów: {full_total}')
print(f'Średni wiek: {full_avg_age:.2f}')
print(f'Wskaźnik przeżycia: {full_survival_rate:.4f}')
```
