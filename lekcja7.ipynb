```python
wagi = [70, 85, 62, 95, 78]
wzrosty = [1.75, 1.80, 1.65, 1.78, 1.72]

# Dla każdego pacjenta ( for i in range(len(wagi)) ) oblicz bmi = wagi[i] / wzrosty[i] ** 2 i
# sklasyfikuj:
# < 18.5 → niedowaga
# 18.5–24.9 → norma
# 25.0–29.9 → nadwaga
# ≥ 30 → otyłość


for i in range(len(wagi)):
    bmi=wagi[i] / wzrosty[i] ** 2
    if bmi < 18.5:
        klasyfikacja='niedowaga'
    elif bmi >= 18.5 and bmi < 24.9:
        klasyfikacja='norma'     
    elif bmi >=25 and bmi < 29.9:
        klasyfikacja='nadwaga'
    elif bmi >=30:
        klasyfikacja='otyłość'
    print(f"pacjent {i}, klasyfikacja:{klasyfikacja}")
```

```python
# Znajdź wszystkie liczby pierwsze w zakresie 2–100. 
# Zewnętrzna pętla: for n in range(2, 101) .
# Wewnętrzna: sprawdź dzielniki od 2 do int(n ** 0.5) + 1 . Zmienna czy_pierwsza = True – ustaw
# False i break , gdy znajdziesz dzielnik. Po pętli wewnętrznej, jeśli czy_pierwsza , dodaj do listy. Wypisz
# listę i jej długość (powinno być 25 liczb).

pierwsze = []
for n in range(2, 101):
    czy_pierwsza = True
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            czy_pierwsza = False
            break
        if czy_pierwsza:
            pierwsze.append(n)
        # print(n)
        # print(check)

print(f"Liczby pierwsze: {pierwsze}")
print(f"Długość listy: {len(pierwsze)}")
```

```python

# Dla każdego miesiąca ( i from 1 to 11 ) porównaj z poprzednim i oblicz procentową zmianę. Wypisz tylko
# miesiące, w których odnotowano wzrost, w formacie: Miesiąc 2: wzrost z 1200 do 1350 (+12.50%) . Na
# końcu wypisz: liczbę miesięcy ze wzrostem, liczbę ze spadkiem i średni dodatni wzrost (tylko po miesiącach
# ze wzrostem).


sprzedaz = [1200, 1350, 1290, 1480, 1520, 1510, 1680, 1720, 1700, 1890, 1950, 2100]

wzrosty_liczba = 0
spadki_liczba = 0
suma_procentow_wzrostu = 0

for i in range(1, len(sprzedaz)):
    poprzednia = sprzedaz[i - 1]
    aktualna = sprzedaz[i]
    
    zmiana = ((aktualna - poprzednia) / poprzednia) * 100
    
    if aktualna > poprzednia:
        wzrosty_liczba += 1
        suma_procentow_wzrostu += zmiana
        print(f"Miesiąc {i + 1}: wzrost z {poprzednia} do {aktualna} (+{zmiana:.2f}%)")
    elif aktualna < poprzednia:
        spadki_liczba += 1

sredni_wzrost = suma_procentow_wzrostu / wzrosty_liczba if wzrosty_liczba > 0 else 0

print(f"\nLiczba miesięcy ze wzrostem: {wzrosty_liczba}")
print(f"Liczba miesięcy ze spadkiem: {spadki_liczba}")
print(f"Średni dodatni wzrost: {sredni_wzrost:.2f}%")
```

```python
wzrost = [165, 170, 175, 180, 185]
waga = [60, 65, 70, 75, 80]

# Oblicz współczynnik korelacji Pearsona:
# 1. Oblicz średnie: sr_x = sum(wzrost) / len(wzrost) , sr_y = sum(waga) / len(waga) .
# 2. W jednej pętli for i in range(len(wzrost)) policz trzy sumy: sum_xy , sum_x2 , sum_y2 :
# sum_xy += (wzrost[i] - sr_x) * (waga[i] - sr_y)
# sum_x2 += (wzrost[i] - sr_x) ** 2
# sum_y2 += (waga[i] - sr_y) ** 2
# 1. r = sum_xy / (sum_x2 * sum_y2) ** 0.5 .
# 2. Wypisz r z dokładnością do 4 miejsc. Oczekiwany wynik: r = 1.0000 (idealna korelacja dodatnia).

sr_x = sum(wzrost) / len(wzrost)
sr_y = sum(waga) / len(waga)

sum_xy = 0
sum_x2 = 0
sum_y2 = 0

for i in range(len(wzrost)):
    sum_xy += (wzrost[i] - sr_x) * (waga[i] - sr_y)
    sum_x2 += (wzrost[i] - sr_x) ** 2
    sum_y2 += (waga[i] - sr_y) ** 2

r = sum_xy / (sum_x2 * sum_y2) ** 0.5

print(f"r = {r:.4f}")
```

```python
wyniki = [78, 82, 85, 88, 90, 92, 95, 97, 99, 45, 101, 103, 105, 180]

# Oblicz outliers:
# 1. posort = sorted(wyniki) .
# 2. Wyznacz Q1 i Q3 „ręcznie":
# Q1 = posort[len(posort) // 4] (25-ty percentyl – uproszczenie)
# Q3 = posort[3 * len(posort) // 4] (75-ty percentyl)
# 1. IQR = Q3 - Q1 , granice = [Q1 - 1.5 * IQR, Q3 + 1.5 * IQR] .

# 2. List comprehension – wyłuskaj outliers: [x for x in wyniki if x < dolna or x > gorna] .
# 3. Wypisz Q1, Q3, IQR, granice i listę outlierów.


posort = sorted(wyniki)

Q1 = posort[len(posort) // 4]
Q3 = posort[3 * len(posort) // 4]

IQR = Q3 - Q1
granice = [Q1 - 1.5 * IQR, Q3 + 1.5 * IQR]
dolna, gorna = granice
print(f"granice: {granice}")
outliers = [x for x in wyniki if x < dolna or x > gorna]

print(f"Q1 = {Q1}")
print(f"Q3 = {Q3}")
print(f"IQR = {IQR}")
print(f"Granice: {granice}")
print(f"Outliers: {outliers}")
```
